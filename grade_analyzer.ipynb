```python
import numpy as np
grades = np.array([85, 45, 92, 38,
                   78, 55, 96, 42,
                   88, 72, 61, 49])
print("=== Grades Analyzer ===")
total_student = np.size(grades)
average_grade = np.mean(grades)
highest_grade = np.max(grades)
lowest_grade = np.min(grades)
sort_grade = np.sort(grades)[::-1]
filter_grade = grades >= 50
pass_grade = grades[filter_grade]
filter_grade = grades < 50
fail_grade = grades[filter_grade]
count_pass = len(pass_grade)
count_fail = len(fail_grade)
pass_pencentage = (count_pass/total_student ) * 100

print("Total Student:", total_student)
print("Average Grade:", average_grade)
print("Highest Grade:", highest_grade)
print("lowest Grade:", lowest_grade)
print("Sorted grades [Highest to Lowest]:", sort_grade)
print("Passed:", count_pass)
print("Failed:", count_fail)
print("Pass grades:", pass_grade)
print("Fail grades:", fail_grade)
print("pencentage pass:", pass_pencentage)
print("="*20)

```
